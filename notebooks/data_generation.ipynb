```python
import pandas as pd
import random
from conneg_functions import execute_to_df, generate_sparql

# Define templates and patterns based on the notebook analysis
TEMPLATES = [
    {
        "template_index": 1,
        "template": "List observatories with their names, countries, and locations.",
        "sparql_wikidata": "SELECT ?observatory ?country ?location WHERE { ... }",
        "sparql_dbpedia18": "SELECT ?observatory ?country ?location WHERE { ... }",
    },
    {
        "template_index": 2,
        "template": "Count sampling events for observatories.",
        "sparql_wikidata": "SELECT ?observatory (COUNT(?event) AS ?count) WHERE { ... }",
        "sparql_dbpedia18": "SELECT ?observatory (COUNT(?event) AS ?count) WHERE { ... }",
    },
    # Add more templates as needed
]


# Generate synthetic data
def generate_synthetic_data(num_samples=100):
    data = []
    for _ in range(num_samples):
        template = random.choice(TEMPLATES)
        nnqt_question = f"NNQT: {template['template']}"
        uid = f"UID-{random.randint(1000, 9999)}"
        subgraph = f"Subgraph-{random.randint(1, 10)}"
        paraphrased_question = f"Paraphrased: {template['template']}"
        question = f"Question: {template['template']}"

        data.append(
            {
                "NNQT_question": nnqt_question,
                "uid": uid,
                "subgraph": subgraph,
                "template_index": template["template_index"],
                "question": question,
                "sparql_wikidata": template["sparql_wikidata"],
                "sparql_dbpedia18": template["sparql_dbpedia18"],
                "template": template["template"],
                "paraphrased_question": paraphrased_question,
            }
        )
    return pd.DataFrame(data)


# Save synthetic data to CSV
def save_to_csv(dataframe, output_path="synthetic_data.csv"):
    dataframe.to_csv(output_path, index=False)
    print(f"Synthetic data saved to {output_path}")


if __name__ == "__main__":
    synthetic_data = generate_synthetic_data(num_samples=100)
    save_to_csv(synthetic_data)
```
